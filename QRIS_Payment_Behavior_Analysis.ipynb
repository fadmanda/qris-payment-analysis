## Project Overview

This project explores the relationship between perceived ease of use, perceived benefits, perceived security, and purchase decision scores using a synthetic dataset.

The dataset contains 500 simulated records created for analytical practice. It does not contain real survey responses and does not represent actual QRIS users or university students.


## 1. Synthetic Data Generation

Generate a simulated dataset containing demographic characteristics, QRIS usage behavior, Likert-scale responses, and calculated scores.


```python

# QRIS Payment Behavior Analysis
# Synthetic Portfolio Project

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Python siap digunakan!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python

import pandas as pd
import numpy as np

# Agar hasil simulasi bisa dibuat ulang secara konsisten
np.random.seed(42)

# Jumlah data simulasi
n = 500

# Membuat data dasar responden
df = pd.DataFrame({
    "respondent_id": [f"R{i:03d}" for i in range(1, n + 1)],
    "age": np.random.choice(
        [18, 19, 20, 21, 22, 23, 24, 25],
        size=n,
        p=[0.08, 0.14, 0.20, 0.22, 0.17, 0.10, 0.06, 0.03]
    ),
    "occupation": np.random.choice(
        ["Student", "Employee", "Self-employed", "Other"],
        size=n,
        p=[0.75, 0.15, 0.06, 0.04]
    ),
    "monthly_income": np.random.choice(
        ["Below 1M", "1M-2.9M", "3M-4.9M", "5M or more"],
        size=n,
        p=[0.35, 0.35, 0.20, 0.10]
    ),
    "qris_experience": np.random.choice(
        ["Yes", "No"],
        size=n,
        p=[0.85, 0.15]
    ),
    "qris_frequency": np.random.choice(
        ["Never", "Rarely", "Sometimes", "Often", "Very often"],
        size=n,
        p=[0.15, 0.10, 0.25, 0.30, 0.20]
    ),
    "preferred_payment": np.random.choice(
        ["QRIS", "Cash", "Debit/Credit Card", "E-wallet without QRIS", "Bank Transfer"],
        size=n,
        p=[0.45, 0.20, 0.10, 0.15, 0.10]
    ),
    "main_purchase_channel": np.random.choice(
        ["Offline", "Online", "Both equally"],
        size=n,
        p=[0.40, 0.30, 0.30]
    )
})

# Membuat skor Likert sintetis (1–5)
# Angka ini adalah simulasi, bukan jawaban survei nyata.
ease_base = np.random.normal(3.7, 0.7, n)
benefit_base = np.random.normal(3.8, 0.7, n)
security_base = np.random.normal(3.3, 0.8, n)

# Membuat skor keputusan transaksi yang berkaitan secara simulatif
decision_base = (
    0.35 * ease_base
    + 0.35 * benefit_base
    + 0.15 * security_base
    + np.random.normal(0, 0.65, n)
)

def make_likert(values):
    return np.clip(np.rint(values), 1, 5).astype(int)

# Kemudahan penggunaan
df["ease_1"] = make_likert(ease_base + np.random.normal(0, 0.45, n))
df["ease_2"] = make_likert(ease_base + np.random.normal(0, 0.45, n))
df["ease_3"] = make_likert(ease_base + np.random.normal(0, 0.45, n))

# Manfaat yang dirasakan
df["benefit_1"] = make_likert(benefit_base + np.random.normal(0, 0.45, n))
df["benefit_2"] = make_likert(benefit_base + np.random.normal(0, 0.45, n))
df["benefit_3"] = make_likert(benefit_base + np.random.normal(0, 0.45, n))

# Persepsi keamanan
df["security_1"] = make_likert(security_base + np.random.normal(0, 0.50, n))
df["security_2"] = make_likert(security_base + np.random.normal(0, 0.50, n))
df["security_3"] = make_likert(security_base + np.random.normal(0, 0.50, n))

# Indikator keputusan transaksi
df["decision_1"] = make_likert(decision_base + np.random.normal(0, 0.45, n))
df["decision_2"] = make_likert(decision_base + np.random.normal(0, 0.45, n))
df["decision_3"] = make_likert(decision_base + np.random.normal(0, 0.45, n))

# Membuat skor rata-rata untuk setiap konstruk
df["ease_score"] = df[["ease_1", "ease_2", "ease_3"]].mean(axis=1).round(2)
df["benefit_score"] = df[["benefit_1", "benefit_2", "benefit_3"]].mean(axis=1).round(2)
df["security_score"] = df[["security_1", "security_2", "security_3"]].mean(axis=1).round(2)
df["decision_score"] = df[["decision_1", "decision_2", "decision_3"]].mean(axis=1).round(2)

print("Dataset berhasil dibuat!")
print("Ukuran dataset:", df.shape)
display(df.head())
```

## 2. Data Validation

Inspect the dataset structure, data types, missing values, duplicate records, descriptive statistics, and category distributions.


```python

# 1. Memeriksa ukuran dan struktur dataset
print("Ukuran dataset:", df.shape)

print("\nInformasi kolom:")
df.info()

# 2. Memeriksa nilai kosong
print("\nJumlah missing value per kolom:")
display(df.isnull().sum())

# 3. Memeriksa baris duplikat
print("\nJumlah baris duplikat:", df.duplicated().sum())

# 4. Statistik deskriptif
print("\nStatistik deskriptif:")
display(df.describe().round(2))

# 5. Memeriksa kategori pada kolom utama
print("\nDistribusi pengalaman menggunakan QRIS:")
display(df["qris_experience"].value_counts())

print("\nDistribusi frekuensi penggunaan QRIS:")
display(df["qris_frequency"].value_counts())

print("\nDistribusi metode pembayaran pilihan:")
display(df["preferred_payment"].value_counts())
```

### 2.1 QRIS Data Consistency

Ensure that respondents simulated as never having used QRIS are assigned a usage frequency of "Never".


```python

# Menjaga konsistensi antara pengalaman dan frekuensi penggunaan QRIS

# Responden yang belum pernah menggunakan QRIS
df.loc[df["qris_experience"] == "No", "qris_frequency"] = "Never"

# Responden yang pernah menggunakan QRIS
mask = df["qris_experience"] == "Yes"

df.loc[mask, "qris_frequency"] = np.random.choice(
    ["Rarely", "Sometimes", "Often", "Very often"],
    size=mask.sum(),
    p=[0.12, 0.30, 0.36, 0.22]
)

# Periksa kembali konsistensinya
print("Pengalaman QRIS vs frekuensi penggunaan:")
display(pd.crosstab(df["qris_experience"], df["qris_frequency"]))

print("\nJumlah responden yang belum pernah menggunakan QRIS:")
print((df["qris_experience"] == "No").sum())

print("\nJumlah responden yang belum pernah menggunakan QRIS tetapi frekuensinya bukan Never:")
print(
    (
        (df["qris_experience"] == "No")
        & (df["qris_frequency"] != "Never")
    ).sum()
)
```

## 3. Exploratory Data Analysis (EDA)

Load the saved dataset and inspect its structure before conducting exploratory analysis.

Summarize the main synthetic scores using descriptive statistics, including the mean, standard deviation, and quartiles.

## 4. Correlation Analysis

Examine the linear relationships between the main synthetic scores using a correlation matrix and heatmap.


```python

import matplotlib.pyplot as plt
import pandas as pd

score_columns = [
    "ease_score",
    "benefit_score",
    "security_score",
    "decision_score"
]

# Menghitung korelasi antar-skor
corr_matrix = df[score_columns].corr()

print("Correlation matrix:")
display(corr_matrix.round(2))

# Membuat visualisasi korelasi
plt.figure(figsize=(8, 6))

plt.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlation coefficient")

plt.xticks(
    range(len(score_columns)),
    score_columns,
    rotation=45,
    ha="right"
)

plt.yticks(range(len(score_columns)), score_columns)

# Menampilkan angka korelasi pada setiap sel
for i in range(len(score_columns)):
    for j in range(len(score_columns)):
        plt.text(
            j, i,
            f"{corr_matrix.iloc[i, j]:.2f}",
            ha="center",
            va="center"
        )

plt.title("Correlation Matrix of QRIS Synthetic Dataset")
plt.tight_layout()
plt.show()
```

## 5. Multiple Linear Regression

Fit an ordinary least squares (OLS) regression model to examine the associations between the predictor scores and the synthetic purchase decision score.


```python

import statsmodels.api as sm

# Variabel prediktor dan target
X = df[["ease_score", "benefit_score", "security_score"]]
y = df["decision_score"]

# Menambahkan konstanta untuk intercept
X = sm.add_constant(X)

# Membuat dan menjalankan model regresi linear
model = sm.OLS(y, X).fit()

# Menampilkan ringkasan hasil
print(model.summary())
```

## 6. Model Diagnostics

Use residual plots, a Q-Q plot, and a residual histogram to visually assess regression model assumptions.


```python

import matplotlib.pyplot as plt
import statsmodels.api as sm

# Mengambil nilai prediksi dan residual dari model regresi
predicted = model.fittedvalues
residuals = model.resid

# Grafik 1: Residual vs Fitted Values
plt.figure(figsize=(7, 4))
plt.scatter(predicted, residuals, alpha=0.6)
plt.axhline(y=0, linestyle="--")
plt.title("Residuals vs Fitted Values")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.tight_layout()
plt.show()

# Grafik 2: Q-Q Plot
sm.qqplot(residuals, line="45", fit=True)
plt.title("Q-Q Plot of Regression Residuals")
plt.tight_layout()
plt.show()

# Grafik 3: Histogram Residual
plt.figure(figsize=(7, 4))
plt.hist(residuals, bins=20, edgecolor="black")
plt.title("Distribution of Regression Residuals")
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

### 6.1 Multicollinearity Check

Calculate the variance inflation factor (VIF) for each predictor to assess multicollinearity in the model.


```python

from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd

# Variabel prediktor saja, tanpa target
X_vif = df[["ease_score", "benefit_score", "security_score"]]

# Tambahkan konstanta
X_vif = sm.add_constant(X_vif)

# Hitung VIF
vif_table = pd.DataFrame({
    "Variable": X_vif.columns,
    "VIF": [
        variance_inflation_factor(X_vif.values, i)
        for i in range(X_vif.shape[1])
    ]
})

display(vif_table.round(3))
```

## 7. Regression Results Summary

Combine regression coefficients, p-values, VIF values, and correlations with the synthetic decision score into one summary table.


```python

# Membuat ringkasan koefisien regresi dan VIF

from statsmodels.stats.outliers_influence import variance_inflation_factor
import pandas as pd
import statsmodels.api as sm

# Variabel prediktor
predictors = ["ease_score", "benefit_score", "security_score"]

# Menghitung VIF
X_vif = sm.add_constant(df[predictors])

vif_values = {
    X_vif.columns[i]: variance_inflation_factor(X_vif.values, i)
    for i in range(X_vif.shape[1])
}

# Mengambil koefisien dan p-value dari model regresi
regression_summary = pd.DataFrame({
    "Variable": predictors,
    "Coefficient": [model.params[var] for var in predictors],
    "P_value": [model.pvalues[var] for var in predictors],
    "VIF": [vif_values[var] for var in predictors]
})

regression_summary["Correlation_with_decision"] = [
    df[var].corr(df["decision_score"])
    for var in predictors
]

display(regression_summary.round(3))

print("R-squared:", round(model.rsquared, 3))
print("Adjusted R-squared:", round(model.rsquared_adj, 3))
```

### 7.1 Regression Coefficients and Confidence Intervals

Visualize the estimated regression coefficients and their 95% confidence intervals.


```python

import pandas as pd
import matplotlib.pyplot as plt

# Ambil koefisien dan confidence interval dari model regresi
variables = ["ease_score", "benefit_score", "security_score"]

coef = model.params[variables]
conf = model.conf_int().loc[variables]

results = pd.DataFrame({
    "Variable": variables,
    "Coefficient": coef.values,
    "Lower_CI": conf[0].values,
    "Upper_CI": conf[1].values
})

# Urutkan koefisien agar mudah dibaca
results = results.sort_values("Coefficient")

# Buat grafik koefisien dengan 95% confidence interval
plt.figure(figsize=(8, 5))

plt.errorbar(
    x=results["Coefficient"],
    y=results["Variable"],
    xerr=[
        results["Coefficient"] - results["Lower_CI"],
        results["Upper_CI"] - results["Coefficient"]
    ],
    fmt="o",
    capsize=5
)

plt.axvline(x=0, linestyle="--")
plt.title("Regression Coefficients with 95% Confidence Intervals")
plt.xlabel("Regression Coefficient")
plt.ylabel("Predictor")
plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()

display(results.round(3))
```
